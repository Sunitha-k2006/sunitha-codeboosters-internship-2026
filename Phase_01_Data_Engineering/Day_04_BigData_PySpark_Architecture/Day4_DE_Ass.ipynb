|   df_filtered = df. filter(df.age > 18)    | df_filtered.show() |
|    |                    |


```python
'''
  3. Simple Python Code: Top 5 Cities by Revenue

'''

import pandas as pd

# Sample sales data
data = {
    "City": ["Chennai", "Mumbai", "Delhi", "Chennai", "Mumbai"],
    "Revenue": [1000, 2000, 1500, 3000, 2500]
}

df = pd.DataFrame(data)

# Total revenue by city
result = df.groupby("City")["Revenue"].sum()

# Top 5 cities
top5 = result.sort_values(ascending=False).head(5)

print(top5)
```

**5. Medallion Architecture**

      Raw data from source systems.

Silver Layer

      Cleaned and transformed data.

Gold Layer

     Business-ready aggregated data for reporting.

Diagram:

       Source Data

**6. Data Lake vs Data Warehouse**

| Data Lake                                 | Data Warehouse          |
| ----------------------------------------- | ----------------------- |
| Stores raw data                           | Stores processed data   |
| Structured, semi-structured, unstructured | Mostly structured       |
| Flexible and low cost                     | Optimized for reporting |

**Use Data Lake:**
      For storing large amounts of raw data.

**Use Data Warehouse:**
      For business reports, dashboards, and analytics.
